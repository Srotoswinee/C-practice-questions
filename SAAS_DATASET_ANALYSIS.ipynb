```python
import matplotlib.pyplot as plt
import numpy as np

product_sales=df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
print(product_sales)
plt.figure(figsize=(10,5))
product_sales.plot(kind='bar', color='steelblue')
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

discount_profit = df.groupby('Discount')['Profit'].mean().reset_index()
print(discount_profit)


plt.figure(figsize=(10,5))
plt.bar(discount_profit['Discount'].astype(str),
        discount_profit['Profit'],
        color='coral')
plt.title('Average Profit by Discount Level')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.axhline(y=0, color='red', linestyle='--', label='Zero Profit Line')
plt.tight_layout()
plt.show()
```

```python
# At which discount does profit go negative?
discount_profit[discount_profit['Profit'] < 0]
```

```python
import plotly.express as px

product_discount = df.groupby('Product')[['Discount','Profit']].mean().sort_values('Discount', ascending=False).reset_index()

fig = px.bar(product_discount,
             x='Product',
             y='Discount',
             color='Profit',
             hover_data=['Product', 'Discount', 'Profit'],
             title='Average Discount vs Profit by Product',
             color_continuous_scale='RdYlGn')

fig.show()
```

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False).reset_index()

fig = px.bar(region_profit,
             x='Region',
             y='Profit',
             color='Profit',
             hover_data=['Region','Profit'],
             title='Total Profit by Region',
             color_continuous_scale='RdYlGn')

fig.show()
```

```python
industry_profit = df.groupby('Industry')['Profit'].sum().sort_values(ascending=False).reset_index()

fig = px.bar(industry_profit,
             x='Industry',
             y='Profit',
             color='Profit',
             hover_data=['Industry','Profit'],
             title='Total Profit by Industry',
             color_continuous_scale='RdYlGn')

fig.show()
```

```python
segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False).reset_index()

fig = px.bar(segment_profit,
             x='Segment',
             y='Profit',
             color='Profit',
             hover_data=['Segment','Profit'],
             title='Total Profit by Customer Segment',
             color_continuous_scale='RdYlGn')

fig.show()
```
